### Ejercicio 3: Conversor de temperaturas
Haz un programa que convierta de grados **Celsius a Fahrenheit** y de **Fahrenheit a Celsius**.
- Pide al usuario una temperatura y la escala de origen.
- Devuelve la conversión correcta.

Fórmulas:
- °F = (°C × 9/5) + 32
- °C = (°F − 32) × 5/9

````python
# Escribe aquí tu solución
temp = int(input("Introduce la temperatura: "))
tipo = input("Introduce si es Clecius (C) o Fahrenheit (F)").lower()[0]

match tipo:
  case "c":
    farenheit = (temp * 9/5) + 32
    print(f"la temperatura en Farenheit es {farenheit}")
  case "f":
    celcius = (temp - 32) * 5/9
    print(f"la temperatura en Celcius es {celcius}")
  case _:
    print("Te has equivocado")
````

### Ejercicio 9: Agenda de contactos
Crea un programa que permita guardar contactos en un diccionario con la estructura:
`nombre -> teléfono, email`

Debe permitir:
- Añadir un nuevo contacto.
- Buscar un contacto por nombre.
- Mostrar todos los contactos guardados.

````python
# Escribe aquí tu solución

contactos = {"Adrià": [55555, "a@a.com"]}

while True:
  print("1. Añadir nuevo contacto")
  print("2. Buscar contacto por nombre")
  print("3. Mostrar todos los contactos")
  opc = int(input("Introduce tu opcion: "))

  match opc:
    case 1:
      nombre = input("Introduce el nombre: ")
      tlf = int(input("Introduce el numero de telefono: "))
      email = input("Introduce el email del contacto: ")

      contactos[nombre] = [tlf, email]
    case 2:
      nombre = input("Introduce el nombre a buscar: ")
      if nombre in contactos:
        print(f"{nombre} {contactos[nombre]}")
      else:
        print("Usuario no encontrado")
    case 3:
      for con in contactos.keys():
        print(f"{con} ,{contactos[con][0]} {contactos[con][1]}")
    case _:
      break
````

## 9. Herencia en POO
### Teoría
- La **herencia** permite que una clase (hija) reutilice atributos y métodos de otra clase (padre).
- Se usa para evitar repetir código y crear jerarquías de objetos.
- Sintaxis básica: `class Hija(Padre): ...`


````python
# Ejemplo: herencia simple
class Vehiculo:
    def __init__(self, marca):
        self.marca = marca

    def mover(self):
        print(f'El vehículo {self.marca} se mueve')

class Coche(Vehiculo):
    def __init__(self, marca, puertas):
        super().__init__(marca)
        self.puertas = puertas

    def mover(self):
        print(f'El coche {self.marca} avanza sobre ruedas')

v = Vehiculo('Genérico')
v.mover()

c = Coche('Toyota', 4)
c.mover()
````
