````python
# Escribe aquí tu solución

conversion = {
    "D": 1.08,   # Dólares (USD)
    "L": 0.86,   # Libras (GBP)
    "Y": 160.50  # Yenes (JPY)
}

def convertir_euros(cantidad, moneda_destino):
    moneda_destino = moneda_destino.upper()

    if moneda_destino in conversion:
        resultado = cantidad * conversion[moneda_destino]
        return resultado
    else:
        return None

euros = float(input("Introduce la cantidad en euros (€): "))

print("\nMonedas disponibles:")
print("D - Dólares")
print("L - Libras")
print("Y - Yenes")

opcion = input("Introduce la letra de la moneda destino (D/L/Y): ")

resultado = convertir_euros(euros, opcion)

if resultado is not None:
    print(f"\n{euros} € equivalen a {resultado:.2f} en la moneda seleccionada.")
else:
    print("\nOpción de moneda no válida.")
````

### Ejercicio 13: Filtrado con funciones

- Usa `filter` y `lambda` para obtener los números pares.
- Usa `map` para elevarlos al cuadrado.

````python
numeros = list(range(1, 21))

pares = filter(lambda x: x % 2 == 0, numeros)

cuadrados = map(lambda x: x ** 2, pares)

resultado = list(cuadrados)

print(resultado)
````

### Ejercicio 18: Biblioteca
Crea una clase `Libro` con atributos `título`, `autor` y `año`.
Crea también una clase `Biblioteca` que permita:
- Añadir libros
- Buscar libros por título
- Listar todos los libros

````python
class Libro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano

    def __str__(self):
        return f"'{self.titulo}' de {self.autor} ({self.ano})"


class Biblioteca:
    def __init__(self):
        self.libros = []

    def anadir_libro(self, libro):
        """Añade un objeto Libro a la biblioteca."""
        self.libros.append(libro)
        print(f"Libro '{libro.titulo}' añadido con éxito.")

    def buscar_por_titulo(self, titulo_buscar):
        """Busca un libro por su título (ignorando mayúsculas/minúsculas)."""
        encontrados = []
        for libro in self.libros:
            if titulo_buscar.lower() in libro.titulo.lower():
                encontrados.append(libro)

        if encontrados:
            print(f"\n--- Libros encontrados con '{titulo_buscar}' ---")
            for libro in encontrados:
                print(libro)
        else:
            print(f"\nNo se encontró ningún libro con el título '{titulo_buscar}'.")

    def listar_libros(self):
        """Muestra todos los libros registrados en la biblioteca."""
        if not self.libros:
            print("\nLa biblioteca está vacía.")
        else:
            print("\n--- Catálogo de la Biblioteca ---")
            for i, libro in enumerate(self.libros, 1):
                print(f"{i}. {libro}")




mi_biblioteca = Biblioteca()


libro1 = Libro("Cien años de soledad", "Gabriel García Márquez", 1967)
libro2 = Libro("Don Quijote de la Mancha", "Miguel de Cervantes", 1605)
libro3 = Libro("El Principito", "Antoine de Saint-Exupéry", 1943)


mi_biblioteca.anadir_libro(libro1)
mi_biblioteca.anadir_libro(libro2)
mi_biblioteca.anadir_libro(libro3)


mi_biblioteca.listar_libros()


mi_biblioteca.buscar_por_titulo("Quijote")
````

### Proyecto 1: Gestor de tareas
Crea un programa en consola que permita:
- Añadir tareas
- Listar tareas pendientes
- Marcar tareas como completadas
- Eliminar tareas

````python
class Tarea:
    def __init__(self, nombre, cuerpo, iscompletada=False):
        self.nombre = nombre
        self.cuerpo = cuerpo
        self.iscompletada = iscompletada

    def __str__(self):
        estado = "[✓] Completada" if self.iscompletada else "[ ] Pendiente"
        return f"{estado} | {self.nombre}: {self.cuerpo}"


class GestorTareas:
    def __init__(self):
        self.lista_tareas = []

    def anadir_tarea(self, nombre, cuerpo):
        nueva_tarea = Tarea(nombre, cuerpo)
        self.lista_tareas.append(nueva_tarea)
        print(f"Tarea '{nombre}' añadida con éxito.")

    def listar_pendientes(self):
        pendientes = [t for t in self.lista_tareas if not t.iscompletada]

        if not pendientes:
            print("\nNo hay tareas pendientes.")
            return

        print("\n--- Tareas Pendientes ---")
        for i, tarea in enumerate(self.lista_tareas):
            if not tarea.iscompletada:
                print(f"{i + 1}. {tarea}")

    def listar_todas(self):
        if not self.lista_tareas:
            print("\nLa lista de tareas está vacía.")
            return

        print("\n--- Todas las Tareas ---")
        for i, tarea in enumerate(self.lista_tareas, 1):
            print(f"{i}. {tarea}")

    def completar_tarea(self, indice):
        if 0 <= indice < len(self.lista_tareas):
            self.lista_tareas[indice].iscompletada = True
            print(f"Tarea '{self.lista_tareas[indice].nombre}' marcada como completada.")
        else:
            print("Número de tarea no válido.")

    def eliminar_tarea(self, indice):
        if 0 <= indice < len(self.lista_tareas):
            eliminada = self.lista_tareas.pop(indice)
            print(f"Tarea '{eliminada.nombre}' eliminada.")
        else:
            print("Número de tarea no válido.")


def menu():
    gestor = GestorTareas()

    while True:
        print("\n=== GESTOR DE TAREAS ===")
        print("1. Añadir tarea")
        print("2. Listar tareas pendientes")
        print("3. Listar TODAS las tareas")
        print("4. Marcar tarea como completada")
        print("5. Eliminar tarea")
        print("6. Salir")

        opcion = input("Selecciona una opción (1-6): ").strip()

        if opcion == "1":
            nombre = input("Nombre de la tarea: ")
            cuerpo = input("Descripción/Cuerpo: ")
            gestor.anadir_tarea(nombre, cuerpo)

        elif opcion == "2":
            gestor.listar_pendientes()

        elif opcion == "3":
            gestor.listar_todas()

        elif opcion == "4":
            gestor.listar_todas()
            try:
                num = int(input("Introduce el número de la tarea a completar: ")) - 1
                gestor.completar_tarea(num)
            except ValueError:
                print("Por favor, introduce un número válido.")

        elif opcion == "5":
            gestor.listar_todas()
            try:
                num = int(input("Introduce el número de la tarea a eliminar: ")) - 1
                gestor.eliminar_tarea(num)
            except ValueError:
                print("Por favor, introduce un número válido.")

        elif opcion == "6":
            print("¡Hasta luego!")
            break

        else:
            print("Opción no válida. Inténtalo de nuevo.")

menu()
````

### Proyecto 3: Analizador de texto
Escribe un programa que lea un archivo `.txt` y muestre:
- Palabras más frecuentes
- Número total de frases
- Longitud media de las palabras

````python
import re
from collections import Counter

def analizar_texto(ruta_archivo):
    try:
        with open(ruta_archivo, 'r', encoding='utf-8') as f:
            contenido = f.read()

        if not contenido.strip():
            print("El archivo está vacío.")
            return

        frases = [f for f in re.split(r'[.!?]+', contenido) if f.strip()]
        total_frases = len(frases)

        palabras = re.findall(r'\b\w+\b', contenido.lower())
        total_palabras = len(palabras)

        if total_palabras == 0:
            print("No se encontraron palabras en el archivo.")
            return

        suma_longitudes = sum(len(palabra) for palabra in palabras)
        longitud_media = suma_longitudes / total_palabras

        conteo_palabras = Counter(palabras)
        palabras_comunes = conteo_palabras.most_common(5)

        print("=== ANÁLISIS DEL TEXTO ===")
        print(f"• Número total de frases: {total_frases}")
        print(f"• Longitud media de las palabras: {longitud_media:.2f} caracteres")
        print("\n• Palabras más frecuentes (Top 5):")
        for palabra, frecuencia in palabras_comunes:
            print(f"  - '{palabra}': {frecuencia} veces")

    except FileNotFoundError:
        print(f"Error: El archivo '{ruta_archivo}' no existe.")

with open("texto_ejemplo.txt", "w", encoding="utf-8") as f:
    f.write("¡Hola! Este es un procesador de texto en Python. "
            "Python es un lenguaje genial y muy popular. "
            "¿Te gusta programar en Python?")

analizar_texto("texto_ejemplo.txt")
````

## 9. Herencia en POO
### Teoría
- La **herencia** permite que una clase (hija) reutilice atributos y métodos de otra clase (padre).
- Se usa para evitar repetir código y crear jerarquías de objetos.
- Sintaxis básica: `class Hija(Padre): ...`


````python
# Ejemplo: herencia simple
class Vehiculo:
    def __init__(self, marca):
        self.marca = marca

    def mover(self):
        print(f'El vehículo {self.marca} se mueve')

class Coche(Vehiculo):
    def __init__(self, marca, puertas):
        super().__init__(marca)
        self.puertas = puertas

    def mover(self):
        print(f'El coche {self.marca} avanza sobre ruedas')

v = Vehiculo('Genérico')
v.mover()

c = Coche('Toyota', 4)
c.mover()
````
